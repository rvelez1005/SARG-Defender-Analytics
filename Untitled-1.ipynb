# We want to scrape FBREF for defender statistics (defensive actions & miscellaneous stats that apply to defenders)

```python
import pandas as pd

url = "https://fbref.com/en/comps/Big5/2023-2024/defense/players/2023-2024-Big-5-European-Leagues-Stats"

tables = pd.read_html(url)

df = tables[0]

df.columns = df.columns.droplevel(0) 

df = df.rename(columns=lambda x: x if not x.startswith("Unnamed") else " ")

df = df.iloc[:, :-1]

df = df[~df["Pos"].str.contains("FW", na=False)]
df = df[~df["Pos"].str.contains("GK", na=False)]

df = df[df["Player"] != "Player"]  # Remove rows where "Player" column repeats the header

df = df.reset_index(drop=True)


print(df)

```

# Now, we do the same for the FBREF page that contains the Miscellaneous stats.

```python
misc_url = "https://fbref.com/en/comps/Big5/2023-2024/misc/players/2023-2024-Big-5-European-Leagues-Stats"

misc_tables = pd.read_html(misc_url)

misc_df = misc_tables[0]

misc_df.columns = misc_df.columns.droplevel(0) 

misc_df = misc_df.rename(columns=lambda x: x if not x.startswith("Unnamed") else " ")

misc_df = misc_df.iloc[:, :-1]

misc_df = misc_df[~misc_df["Pos"].str.contains("FW", na=False)]
misc_df = misc_df[~misc_df["Pos"].str.contains("GK", na=False)]

misc_df = misc_df[misc_df["Player"] != "Player"]  # Remove rows where "Player" column repeats the header

misc_df = misc_df.reset_index(drop=True)


print(misc_df)
```

# Now, we want to process the Defensive & Miscellaneous Stats to get rid of irrelevant columns:
- Player (Name)
- Nation
- Squad
- Comp
- Age
- Born
- PKs Won
- Fouls Drawn

```python
df.drop(columns=["Player", "Nation", "Squad", "Comp", "Age", "Born"], inplace=True)
misc_df.drop(columns=["Player", "Nation", "Squad", "Comp", "Age", "Born", "PKwon", "Fld"], inplace=True)
```

```python
# Sanity Check to make sure the expected rows, cols match the actual rows, cols

print(df.shape)
print(misc_df.shape)
```

```python
df_combined = pd.concat([df, misc_df], axis=1)
print(df_combined.shape)
```

```python
df_combined.to_csv("Combined_table.csv", index=False)
```
